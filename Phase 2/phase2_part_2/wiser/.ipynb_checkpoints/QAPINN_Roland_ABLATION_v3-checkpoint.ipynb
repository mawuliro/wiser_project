# QAPINN Ablation Grid 


## Two things, both important

**1. Deep classical body (≈1850 params).** The project plan says replace *the first hidden layer* of
the Phase-1 PINN, the other seven were meant to stay.

**2. Ablation grid moves to ν = 0.05.** This is a timeline decision. At ν=0.01/π each config needs
shock-focused sampling plus a time-marching curriculum (~2 h each × 14 configs = days). At ν=0.05 the
same architecture reaches ~4e-3 in 15–30 min with no curriculum, and every ablation axis is measured
just as cleanly. **The hard viscosity is already covered by Djabon's hero run**, so nothing is lost.

## 2. Imports & configuration

```python
import os, time, math, json
import numpy as np
import torch
import torch.nn as nn
import pennylane as qml
from scipy.stats import qmc
import matplotlib.pyplot as plt

SEED = 1234
np.random.seed(SEED); torch.manual_seed(SEED)

DTYPE = torch.float32
torch.set_default_dtype(DTYPE)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

CKPT_ROOT = "qapinn_runs"
for sub in ["checkpoints", "results", "figures", "models"]:
    os.makedirs(os.path.join(CKPT_ROOT, sub), exist_ok=True)

N_U = 120          # IC+BC points
print(f"[cfg] torch {torch.__version__} | pennylane {qml.__version__} | {device} | {DTYPE}")
```

## 3. The two PDEs + training-data maker
Both live on $x\in[-1,1],\ t\in[0,1]$ with zero boundary conditions, so all code is shared:

- **Burgers**: $u_t + u\,u_x - \nu\,u_{xx} = 0$, $u(0,x)=-\sin(\pi x)$. Tier 1 uses $\nu=0.05$
  (moderate front, resolvable); Tier 2 uses the hard $\nu=0.01/\pi$ (sharp shock).
- **Heat**: $u_t - \alpha\,u_{xx} = 0$, $u(0,x)=\sin(\pi x)$, exact solution
  $\sin(\pi x)e^{-\alpha\pi^2 t}$ with $\alpha=0.1$ (nice dynamic range over $t\in[0,1]$).


```python
NU_EASY  = 0.05          # Tier-1 Burgers viscosity
NU_HARD  = 0.01/np.pi    # Tier-2 Burgers viscosity (Raissi/Shah benchmark)
ALPHA    = 0.10          # Heat diffusivity

def initial_condition(pde, x):
    return -np.sin(np.pi*x) if pde == "burgers" else np.sin(np.pi*x)

def make_training_data(pde, n_f, tmax=1.0, shock_frac=0.0, seed=SEED):
    """Returns torch tensors (x_u,t_u,u_u, x_f,t_f) on the right device/dtype."""
    rng = np.random.default_rng(seed)
    # initial-condition points
    n_ic = N_U // 2
    x_ic = rng.uniform(-1, 1, n_ic); t_ic = np.zeros(n_ic)
    u_ic = initial_condition(pde, x_ic)
    # boundary points
    n_bc = N_U - n_ic
    t_bc = rng.uniform(0, tmax, n_bc)
    x_bc = np.where(rng.random(n_bc) < 0.5, -1.0, 1.0)
    u_bc = np.zeros(n_bc)
    x_u = np.concatenate([x_ic, x_bc]).reshape(-1, 1)
    t_u = np.concatenate([t_ic, t_bc]).reshape(-1, 1)
    u_u = np.concatenate([u_ic, u_bc]).reshape(-1, 1)
    # collocation: LHS base + optional shock-focused extra (Burgers only)
    n_sh = int(shock_frac * n_f) if pde == "burgers" else 0
    n_lhs = n_f - n_sh
    lhs = qmc.LatinHypercube(d=2, seed=seed)
    s = lhs.random(n_lhs)
    x_f1 = -1.0 + 2.0*s[:, 0]; t_f1 = tmax*s[:, 1]
    x_f2 = np.clip(rng.normal(0.0, 0.15, n_sh), -1, 1)
    t_f2 = rng.uniform(0.25*tmax, tmax, n_sh)
    x_f = np.concatenate([x_f1, x_f2]).reshape(-1, 1)
    t_f = np.concatenate([t_f1, t_f2]).reshape(-1, 1)
    T = lambda a: torch.tensor(a, dtype=DTYPE, device=device)
    return T(x_u), T(t_u), T(u_u), T(x_f), T(t_f)

print("[ok] data maker ready")
```

## 4. Reference solutions (exact, cached)
- Heat: closed form.
- Burgers (any ν): the Phase-1 Cole–Hopf quadrature.

Each reference grid (256×256) is computed once and cached to disk; `relative_l2` measures
$\|\hat u - u\| / \|u\|$ against it.

```python
def burgers_analytical(x_arr, t_arr, nu, n_quad=4000):
    x_arr = np.atleast_1d(np.asarray(x_arr, float)); t_arr = np.atleast_1d(np.asarray(t_arr, float))
    u = np.zeros((len(t_arr), len(x_arr)))
    y = np.linspace(-3.0, 3.0, n_quad)
    cos_term = np.cos(np.pi*y) / (2.0*np.pi*nu)
    for i, ti in enumerate(t_arr):
        if ti < 1e-12:
            u[i, :] = -np.sin(np.pi*x_arr); continue
        F = cos_term.reshape(1, -1) + (x_arr.reshape(-1, 1) - y.reshape(1, -1))**2/(4.0*nu*ti)
        ml = np.max(-F, axis=1, keepdims=True); w = np.exp(-F - ml)
        num = np.sum((x_arr.reshape(-1, 1) - y.reshape(1, -1))/ti * w, axis=1)
        den = np.sum(w, axis=1)
        u[i, :] = np.where(den > 0, num/den, 0.0)
    return u

_REFS = {}
def reference(pde, nu):
    key = (pde, round(float(nu), 8))
    if key in _REFS: return _REFS[key]
    path = os.path.join(CKPT_ROOT, "results", f"ref_{pde}_{nu:.6f}.npz")
    if os.path.exists(path):
        d = np.load(path); out = (d["x"], d["t"], d["U"])
    else:
        x = np.linspace(-1, 1, 256); t = np.linspace(0, 1, 256)
        if pde == "heat":
            U = np.sin(np.pi*x)[None, :] * np.exp(-nu*np.pi**2*t)[:, None]
        else:
            print(f"[ref] computing Cole-Hopf for nu={nu:.5f} (once, ~20s)...")
            U = burgers_analytical(x, t, nu)
        np.savez(path, x=x, t=t, U=U); out = (x, t, U)
    _REFS[key] = out
    return out

@torch.no_grad()
def relative_l2(model, pde, nu):
    x, t, U = reference(pde, nu)
    XX, TT = np.meshgrid(x, t)
    xf = torch.tensor(XX.reshape(-1, 1), dtype=DTYPE, device=device)
    tf = torch.tensor(TT.reshape(-1, 1), dtype=DTYPE, device=device)
    up = model(xf, tf).cpu().numpy().reshape(U.shape)
    return float(np.linalg.norm(up - U)/np.linalg.norm(U)), up

def frozen_ic_baseline(pde, nu):
    """L2 of predicting the initial condition for all t -- the honest 'no-dynamics' bar."""
    x, t, U = reference(pde, nu)
    IC = np.tile(initial_condition(pde, x), (len(t), 1))
    return float(np.linalg.norm(IC - U)/np.linalg.norm(U))

print("[ok] references ready")
```

## 5. The QAPINN model 


```python
def build_qnode(n_qubits, n_layers, entanglement="all", measurement="expval",
                encoding_reps=None):
    dev = qml.device("default.qubit", wires=n_qubits)
    reps = n_layers if encoding_reps is None else encoding_reps
    def entangle():
        if entanglement == "none": return
        if entanglement == "linear":
            for q in range(n_qubits - 1): qml.CNOT(wires=[q, q + 1])
        elif entanglement == "all":
            for a in range(n_qubits):
                for b in range(a + 1, n_qubits): qml.CNOT(wires=[a, b])
    @qml.qnode(dev, interface="torch", diff_method="backprop")
    def circuit(inputs, weights):
        for l in range(n_layers):
            if l < reps:                                  # data re-uploading
                for q in range(n_qubits):
                    qml.RY(inputs[:, q % 2], wires=q)
            for q in range(n_qubits):
                qml.RX(weights[l, q], wires=q)
            entangle()
        if measurement == "expval":
            return [qml.expval(qml.PauliZ(q)) for q in range(n_qubits)]
        return qml.probs(wires=range(n_qubits))
    out_dim = n_qubits if measurement == "expval" else 2**n_qubits
    return circuit, {"weights": (n_layers, n_qubits)}, out_dim


class QAPINN(nn.Module):
    """VQC first layer (re-uploading, trainable input scale) + deep classical body."""
    def __init__(self, n_qubits=3, n_layers=6, entanglement="all", measurement="expval",
                 encoding_reps=None, head_width=20, head_depth=5):
        super().__init__()
        self.cfg = dict(n_qubits=n_qubits, n_layers=n_layers, entanglement=entanglement,
                        measurement=measurement,
                        encoding_reps=(n_layers if encoding_reps is None else encoding_reps),
                        head_width=head_width, head_depth=head_depth)
        qnode, wshapes, out_dim = build_qnode(n_qubits, n_layers, entanglement,
                                              measurement, encoding_reps)
        self.qlayer = qml.qnn.TorchLayer(qnode, wshapes)
        self.in_scale = nn.Parameter(torch.ones(2))
        layers = [nn.Linear(out_dim, head_width), nn.Tanh()]
        for _ in range(head_depth - 1):
            layers += [nn.Linear(head_width, head_width), nn.Tanh()]
        layers += [nn.Linear(head_width, 1)]
        self.head = nn.Sequential(*layers)
        for m in self.head:                                # Xavier init (as Phase 1)
            if isinstance(m, nn.Linear):
                nn.init.xavier_normal_(m.weight); nn.init.zeros_(m.bias)

    def forward(self, x, t):
        inp = torch.cat([x, t], dim=1) * self.in_scale * math.pi
        feat = self.qlayer(inp)
        return self.head(feat)

def count_params(model):
    return sum(p.numel() for p in model.parameters())

_m = QAPINN(3)
print(f"[ok] model ready — 3-qubit example: {count_params(_m)} params "
      f"(vs 101 in the failed v3, 3021 in Phase 1)")
del _m
```

## 6. Physics-informed loss (PDE-aware, exact autograd derivatives)

```python
W_IC = 10.0

def pinn_loss(model, x_u, t_u, u_u, x_f, t_f, pde, nu, w_ic=None):
    w = W_IC if w_ic is None else w_ic
    u_pred = model(x_u, t_u)
    mse_u = torch.mean((u_pred - u_u)**2)
    xf = x_f.clone().requires_grad_(True)
    tf = t_f.clone().requires_grad_(True)
    u = model(xf, tf)
    u_x = torch.autograd.grad(u, xf, torch.ones_like(u), create_graph=True)[0]
    u_t = torch.autograd.grad(u, tf, torch.ones_like(u), create_graph=True)[0]
    u_xx = torch.autograd.grad(u_x, xf, torch.ones_like(u_x), create_graph=True)[0]
    if pde == "burgers":
        f = u_t + u*u_x - nu*u_xx
    else:                              # heat
        f = u_t - nu*u_xx
    mse_f = torch.mean(f**2)
    return w*mse_u + mse_f, mse_u, mse_f
print("[ok] loss ready")
```

## 7. Schedule-based training engine 

```python
def ckpt_path(tag):
    return os.path.join(CKPT_ROOT, "checkpoints", f"{tag}.pt")

def _state_copy(model):
    return {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}

def train_schedule(tag, model, pde, nu, schedule, n_f=2000,
                   ckpt_every=250, log_every=100):
    model.to(device)
    hist = {"iters": [], "loss": [], "mse_u": [], "mse_f": [], "phase": []}
    start_phase, start_it, opt_state = 0, 0, None
    best_loss, best_state = float("inf"), None
    p = ckpt_path(tag)
    if os.path.exists(p):
        ck = torch.load(p, map_location=device, weights_only=False)
        model.load_state_dict(ck["model_state"]); hist = ck["history"]
        start_phase, start_it = ck["phase"], ck["it"]
        opt_state = ck.get("opt_state")
        best_loss = ck.get("best_loss", best_loss); best_state = ck.get("best_state")
        print(f"[resume] '{tag}' at phase {start_phase}, it {start_it}")
    if start_phase >= len(schedule):
        print(f"[done] '{tag}' already finished. Delete its checkpoint to redo.")
        return hist

    cur = {"phase": start_phase, "it": start_it, "opt": None}
    def save():
        tmp = p + ".tmp"
        torch.save({"phase": cur["phase"], "it": cur["it"],
                    "model_state": model.state_dict(),
                    "opt_state": (cur["opt"].state_dict() if cur["opt"] is not None else None),
                    "history": hist, "best_loss": best_loss, "best_state": best_state}, tmp)
        os.replace(tmp, p)

    gstep = len(hist["loss"])
    t0 = time.time()
    try:
        for pi in range(start_phase, len(schedule)):
            ph = schedule[pi]
            cur["phase"] = pi
            x_u, t_u, u_u, x_f, t_f = make_training_data(
                pde, n_f, tmax=ph.get("tmax", 1.0),
                shock_frac=ph.get("shock_frac", 0.0), seed=SEED + 13*pi)
            it0 = start_it if pi == start_phase else 0

            if ph["kind"] == "adam":
                opt = torch.optim.Adam(model.parameters(), lr=ph.get("lr", 8e-3))
                cur["opt"] = opt
                if pi == start_phase and opt_state is not None:
                    try: opt.load_state_dict(opt_state)
                    except Exception: pass
                for it in range(it0, ph["iters"]):
                    cur["it"] = it
                    opt.zero_grad()
                    loss, mu, mf = pinn_loss(model, x_u, t_u, u_u, x_f, t_f, pde=pde, nu=nu)
                    loss.backward(); opt.step()
                    li = loss.item(); gstep += 1
                    hist["iters"].append(gstep); hist["loss"].append(li)
                    hist["mse_u"].append(mu.item()); hist["mse_f"].append(mf.item())
                    hist["phase"].append(pi)
                    if math.isfinite(li) and li < best_loss:
                        best_loss, best_state = li, _state_copy(model)
                    if (it + 1) % log_every == 0:
                        print(f"  [{tag}] P{pi} adam it {it+1:5d}/{ph['iters']} "
                              f"loss={li:.3e} (u={mu.item():.2e}, f={mf.item():.2e}) "
                              f"[{(time.time()-t0)/60:.1f} min]")
                    if (it + 1) % ckpt_every == 0:
                        cur["it"] = it + 1; save()

            else:  # ---- L-BFGS phase, chunked ----
                chunk = ph.get("chunk", 150)
                opt = torch.optim.LBFGS(model.parameters(), max_iter=chunk,
                                        history_size=60, line_search_fn="strong_wolfe",
                                        tolerance_grad=1e-12, tolerance_change=1e-14)
                cur["opt"] = opt
                if pi == start_phase and opt_state is not None:
                    try: opt.load_state_dict(opt_state)
                    except Exception: pass
                n_chunks = math.ceil(ph["iters"]/chunk)
                for c in range(it0 // chunk, n_chunks):
                    def closure():
                        opt.zero_grad()
                        L, _, _ = pinn_loss(model, x_u, t_u, u_u, x_f, t_f, pde=pde, nu=nu)
                        L.backward(); return L
                    opt.step(closure)
                    loss, mu, mf = pinn_loss(model, x_u, t_u, u_u, x_f, t_f, pde=pde, nu=nu)
                    li = loss.item(); gstep += chunk
                    hist["iters"].append(gstep); hist["loss"].append(li)
                    hist["mse_u"].append(mu.item()); hist["mse_f"].append(mf.item())
                    hist["phase"].append(pi)
                    print(f"  [{tag}] P{pi} lbfgs chunk {c+1}/{n_chunks} "
                          f"loss={li:.3e} (u={mu.item():.2e}, f={mf.item():.2e}) "
                          f"[{(time.time()-t0)/60:.1f} min]")
                    if (not math.isfinite(li)) or li > 50*max(best_loss, 1e-12):
                        print(f"  [{tag}] L-BFGS blow-up guard triggered -> "
                              f"restoring best state (loss {best_loss:.3e}) and stopping phase.")
                        if best_state is not None:
                            model.load_state_dict(best_state)
                        break
                    if li < best_loss:
                        best_loss, best_state = li, _state_copy(model)
                    cur["it"] = (c + 1)*chunk; save()

            start_it = 0
            cur["phase"], cur["it"], cur["opt"] = pi + 1, 0, None
            save()

        # keep the best weights seen anywhere in training
        if best_state is not None and hist["loss"] and best_loss < hist["loss"][-1]*0.999:
            model.load_state_dict(best_state)
            print(f"[{tag}] restored overall best state (loss {best_loss:.3e})")
        save()
    except KeyboardInterrupt:
        save()
        print(f"\n[interrupted] '{tag}' checkpointed at phase {cur['phase']}, "
              f"it {cur['it']}. Re-run this cell to continue.")
        raise
    print(f"[finished] '{tag}' total wall time {(time.time()-t0)/60:.1f} min")
    return hist
print("[ok] engine ready")
```

## 8. Results recorder (CSV + saved models + loss histories)

```python
import csv
MASTER_CSV = os.path.join(CKPT_ROOT, "results", "master_results.csv")
FIELDS = ["tag","pde","nu","n_qubits","n_layers","entanglement","measurement",
          "encoding_reps","head_width","head_depth","optimizer","n_iters","params",
          "L2","mse_u","mse_f","final_loss"]

def record_result(tag, model, hist, pde, nu, optimizer="adam+lbfgs"):
    l2, _ = relative_l2(model, pde, nu)
    row = {"tag": tag, "pde": pde, "nu": f"{nu:.6f}", **model.cfg,
           "optimizer": optimizer,
           "n_iters": hist["iters"][-1] if hist["iters"] else 0,
           "params": count_params(model), "L2": f"{l2:.6e}",
           "mse_u": f"{hist['mse_u'][-1]:.6e}" if hist["mse_u"] else "",
           "mse_f": f"{hist['mse_f'][-1]:.6e}" if hist["mse_f"] else "",
           "final_loss": f"{hist['loss'][-1]:.6e}" if hist["loss"] else ""}
    row = {k: v for k, v in row.items() if k in FIELDS}
    exists = os.path.exists(MASTER_CSV)
    with open(MASTER_CSV, "a", newline="") as fh:
        w = csv.DictWriter(fh, fieldnames=FIELDS)
        if not exists: w.writeheader()
        w.writerow(row)
    torch.save(model.state_dict(), os.path.join(CKPT_ROOT, "models", f"{tag}.pt"))
    np.savez(os.path.join(CKPT_ROOT, "results", f"{tag}_history.npz"),
             iters=np.array(hist["iters"]), loss=np.array(hist["loss"]),
             mse_u=np.array(hist["mse_u"]), mse_f=np.array(hist["mse_f"]),
             phase=np.array(hist["phase"]))
    print(f"[record] {tag}: L2={l2:.3e}, params={count_params(model)}")
    return l2
print("[ok] recorder ready")
```

---
# The ablation grid

Every run uses the **identical** validated setup deep body, data re-uploading, trainable input
scale, $W_{IC}=10$, Adam 1000 → L-BFGS 400, $N_f=2000$, Burgers ν=0.05, so that **only the ablated
variable changes**. That is what makes each comparison interpretable.

| Stage | Runs | Ablated axis | Feeds |
|---|---|---|---|
| **B1–B3** | 3,4,5 qubits · linear · ⟨Z⟩ | qubit count (baseline) | WS-A, WS-B, WS-C |
| **B4–B6** | 3,4,5 qubits · **none** · ⟨Z⟩ | entanglement on/off | WS-A, WS-C |
| **B7–B9** | 3,4,5 qubits · linear · **probs** | measurement type | WS-A, WS-C |
| **B10** | 5 qubits · **all-to-all** · ⟨Z⟩ | entanglement strength | WS-A, WS-B |
| **B11–B12** | 4 qubits · linear · ⟨Z⟩ · depth **3** and **9** | ansatz depth | WS-A |
| **B13–B14** | **6, 7** qubits · linear · ⟨Z⟩ | trainability limit | **WS-B** |

```python
# ---- ablation grid definition -------------------------------------------
# Baseline (unless a row overrides it): 6 layers, linear entanglement, <Z>,
# deep body 20x5, Adam 1000 + L-BFGS 400, Nf=2000, Burgers nu=0.05.
PDE, NU = "burgers", NU_EASY

ABLATION_GRID = [
    # tag,        qubits, layers, entanglement, measurement
    ("B1_q3_linear_expval",  3, 6, "linear", "expval"),
    ("B2_q4_linear_expval",  4, 6, "linear", "expval"),
    ("B3_q5_linear_expval",  5, 6, "linear", "expval"),

    ("B4_q3_none_expval",    3, 6, "none",   "expval"),
    ("B5_q4_none_expval",    4, 6, "none",   "expval"),
    ("B6_q5_none_expval",    5, 6, "none",   "expval"),

    ("B7_q3_linear_probs",   3, 6, "linear", "probs"),
    ("B8_q4_linear_probs",   4, 6, "linear", "probs"),
    ("B9_q5_linear_probs",   5, 6, "linear", "probs"),

    ("B10_q5_all_expval",    5, 6, "all",    "expval"),

    ("B11_q4_depth3",        4, 3, "linear", "expval"),
    ("B12_q4_depth9",        4, 9, "linear", "expval"),

    ("B13_q6_linear_expval", 6, 6, "linear", "expval"),
    ("B14_q7_linear_expval", 7, 6, "linear", "expval"),
]

STAGES = {
    "stage1_baseline":      ABLATION_GRID[0:3],
    "stage2_entanglement":  ABLATION_GRID[3:6] + [ABLATION_GRID[9]],
    "stage3_measurement":   ABLATION_GRID[6:9],
    "stage4_depth":         ABLATION_GRID[10:12],
    "stage5_trainability":  ABLATION_GRID[12:14],
}

SCHEDULE = [
    {"kind": "adam",  "lr": 8e-3, "iters": 1000, "tmax": 1.0, "shock_frac": 0.15},
    {"kind": "lbfgs", "iters": 400, "chunk": 100, "tmax": 1.0, "shock_frac": 0.15},
]

B_results = {}

def run_stage(stage_name):
    """Run one stage of the grid. Safe to re-run: finished runs resume/skip."""
    rows = STAGES[stage_name]
    print(f"\n{'='*66}\nSTAGE: {stage_name}  ({len(rows)} runs)\n{'='*66}")
    for tag, nq, nl, ent, meas in rows:
        full_tag = f"{tag}_v3"
        print(f"\n--- {full_tag}: {nq}q, depth {nl}, ent={ent}, meas={meas} ---")
        model = QAPINN(n_qubits=nq, n_layers=nl, entanglement=ent,
                       measurement=meas, head_width=20, head_depth=5)
        print(f"    parameters: {count_params(model)}")
        hist = train_schedule(full_tag, model, PDE, NU, SCHEDULE,
                              n_f=2000, ckpt_every=250, log_every=250)
        l2 = record_result(full_tag, model, hist, PDE, NU)
        B_results[tag] = dict(model=model, hist=hist, l2=l2, n_qubits=nq,
                              n_layers=nl, entanglement=ent, measurement=meas,
                              params=count_params(model))
        print(f"    --> L2 = {l2:.4e}")
    print(f"\n[{stage_name} complete]")

print(f"grid: {len(ABLATION_GRID)} runs across {len(STAGES)} stages")
for s, r in STAGES.items():
    print(f"  {s:22s} {len(r)} runs")
```

### Stage-1 acceptance check

## Gradient-variance probe : the barren-plateau measurement (WS-B)


```python
def gradient_variance(nq, n_samples=25, n_f=400, nl=6, ent="linear", seed=SEED):
    x_u, t_u, u_u, x_f, t_f = make_training_data(PDE, n_f, seed=seed)
    comps = []
    for s in range(n_samples):
        torch.manual_seed(seed + 1000*s)
        m = QAPINN(n_qubits=nq, n_layers=nl, entanglement=ent, measurement="expval",
                   head_width=20, head_depth=5).to(device)
        with torch.no_grad():
            for p_ in m.qlayer.parameters():
                p_.uniform_(0, 2*math.pi)      # random regime the theory describes
        m.zero_grad()
        loss, _, _ = pinn_loss(m, x_u, t_u, u_u, x_f, t_f, pde=PDE, nu=NU)
        loss.backward()
        g = torch.cat([p_.grad.flatten() for p_ in m.qlayer.parameters()])
        comps.append(g.detach().cpu().numpy())
    comps = np.concatenate(comps)
    return float(np.var(comps))

GV_QUBITS = [3, 4, 5, 6, 7]
GV = {}
print(f"{'qubits':>7} {'Var[dL/dtheta]':>17}")
for nq in GV_QUBITS:
    t0 = time.time()
    GV[nq] = gradient_variance(nq)
    print(f"{nq:>7} {GV[nq]:>17.4e}   ({time.time()-t0:.0f}s)")

np.savez(os.path.join(CKPT_ROOT, "results", "B_gradient_variance.npz"),
         qubits=np.array(GV_QUBITS), var=np.array([GV[q] for q in GV_QUBITS]))

ks = sorted(GV)
print("\nsuccessive ratios (consistently < 1 => exponential decay):")
for i in range(1, len(ks)):
    print(f"  Var(q{ks[i]})/Var(q{ks[i-1]}) = {GV[ks[i]]/GV[ks[i-1]]:.3f}")
slope = np.polyfit(ks, np.log([GV[q] for q in ks]), 1)[0]
print(f"\nlog-linear fit:  Var ~ exp({slope:.3f} * n_qubits)")
print("  (clearly negative slope = quantitative barren-plateau signature)")
```

# Neuron-level interpretability (challenge requirement)



```python
# --- activation extraction via forward hooks --------------------------------
import collections

ACT_DIR = os.path.join(CKPT_ROOT, "results", "activations")
os.makedirs(ACT_DIR, exist_ok=True)

@torch.no_grad()
def extract_activations(model, n_grid=200, t_slice=0.5, full_grid=False, n_full=64):
    """Return {layer_name: array} of per-neuron activations.

    L0            = quantum-layer output (measurement results)
    L1..L{D}      = classical body hidden layers (post-tanh)
    Shapes: (n_points, n_neurons) for a 1-D x-slice at fixed t,
            or (n_full*n_full, n_neurons) when full_grid=True.
    """
    model.eval().to(device)
    store = collections.OrderedDict()
    handles = []

    def mk_hook(name):
        def hook(_m, _inp, out):
            o = out.detach().cpu().numpy()
            store[name] = o
        return hook

    # quantum (or classical front) layer output
    if hasattr(model, "qlayer"):
        handles.append(model.qlayer.register_forward_hook(mk_hook("L0_quantum")))
    elif hasattr(model, "front"):
        handles.append(model.front.register_forward_hook(mk_hook("L0_classical_front")))

    # classical body: hook every Tanh (i.e. post-activation of each hidden layer)
    k = 1
    for mod in model.head:
        if isinstance(mod, nn.Tanh):
            handles.append(mod.register_forward_hook(mk_hook(f"L{k}")))
            k += 1

    # evaluation points
    if full_grid:
        xs = np.linspace(-1, 1, n_full); ts = np.linspace(0, 1, n_full)
        XX, TT = np.meshgrid(xs, ts)
        xin = torch.tensor(XX.reshape(-1, 1), dtype=DTYPE, device=device)
        tin = torch.tensor(TT.reshape(-1, 1), dtype=DTYPE, device=device)
        coords = (xs, ts)
    else:
        xs = np.linspace(-1, 1, n_grid)
        xin = torch.tensor(xs.reshape(-1, 1), dtype=DTYPE, device=device)
        tin = torch.full_like(xin, float(t_slice))
        coords = (xs, None)

    model(xin, tin)
    for h in handles:
        h.remove()
    return store, coords


def save_activations(tag, model, t_slice=0.5):
    """Extract and cache activations for one trained model."""
    acts, (xs, _) = extract_activations(model, t_slice=t_slice)
    out = {f"act_{k}": v for k, v in acts.items()}
    out["x"] = xs
    out["t_slice"] = np.array([t_slice])
    np.savez(os.path.join(ACT_DIR, f"{tag}_activations.npz"), **out)
    dims = {k: v.shape[1] for k, v in acts.items()}
    return acts, dims

print("[ok] activation extractor ready")
print("     layer naming: L0 = quantum layer output, L1..L5 = classical body hidden layers")
```

## Figure N1 : Per-neuron activation profiles (the brief's Fig. 3 comparison)



```python
def plot_neuron_profiles(tag, layers=("L1", "L2"), n_show=8, save=True):
    """Small-multiple per-neuron activation profiles. No curve overlap by construction."""
    if tag not in ACTS:
        print(f"[skip] {tag} not available"); return
    acts = ACTS[tag]
    xs = np.linspace(-1, 1, next(iter(acts.values())).shape[0])
    layers = [L for L in layers if L in acts]
    if not layers:
        print(f"[skip] none of the requested layers in {tag}"); return

    ncol = min(n_show, min(acts[L].shape[1] for L in layers))
    fig, axs = plt.subplots(len(layers), ncol,
                            figsize=(1.62*ncol, 2.05*len(layers)),
                            sharex=True, sharey=True, squeeze=False)
    for r, L in enumerate(layers):
        A = acts[L]
        for j in range(ncol):
            ax = axs[r, j]
            ax.plot(xs, A[:, j], color="#14285a", lw=1.5)
            ax.axhline(0, color="#cccccc", lw=0.7, zorder=0)
            ax.set_ylim(-1.15, 1.15)
            ax.grid(alpha=0.18)
            ax.tick_params(labelsize=7.5)
            if r == 0:
                ax.set_title(f"neuron {j}", fontsize=8.5, pad=4)
            if j == 0:
                ax.set_ylabel(f"{L}\nactivation", fontsize=9.5)
            else:
                ax.tick_params(labelleft=False)
            if r == len(layers)-1:
                ax.set_xlabel("$x$", fontsize=9.5)
    fig.suptitle(f"Per-neuron activation profiles — {tag}   (slice at $t=0.5$)",
                 fontsize=12, color="#14285a", weight="bold", y=1.02)
    plt.tight_layout()
    if save:
        plt.savefig(os.path.join(CKPT_ROOT, "figures", f"N1_profiles_{tag}.png"),
                    dpi=160, bbox_inches="tight")
    plt.show()

# baseline 5-qubit model (or whatever is available)
for cand in ["B3_q5_linear_expval", "B2_q4_linear_expval", "B1_q3_linear_expval"]:
    if cand in ACTS:
        plot_neuron_profiles(cand); break
```

## Figure N2 : Entanglement on/off, neuron by neuron



```python
def compare_neuron_profiles(tag_a, tag_b, label_a, label_b, layer="L1",
                            n_show=8, fname="N2_compare"):
    """Stacked (not overlaid) per-neuron comparison between two trained models."""
    if tag_a not in ACTS or tag_b not in ACTS:
        print(f"[skip] need both {tag_a} and {tag_b}"); return
    A, B = ACTS[tag_a].get(layer), ACTS[tag_b].get(layer)
    if A is None or B is None:
        print(f"[skip] layer {layer} missing"); return
    xs = np.linspace(-1, 1, A.shape[0])
    ncol = min(n_show, A.shape[1], B.shape[1])

    fig, axs = plt.subplots(2, ncol, figsize=(1.62*ncol, 4.3),
                            sharex=True, sharey=True, squeeze=False)
    for r, (M, lab, col) in enumerate([(A, label_a, "#14285a"), (B, label_b, "#6a3d9a")]):
        for j in range(ncol):
            ax = axs[r, j]
            ax.plot(xs, M[:, j], color=col, lw=1.5)
            ax.axhline(0, color="#cccccc", lw=0.7, zorder=0)
            ax.set_ylim(-1.15, 1.15); ax.grid(alpha=0.18)
            ax.tick_params(labelsize=7.5)
            if r == 0: ax.set_title(f"neuron {j}", fontsize=8.5, pad=4)
            if j == 0: ax.set_ylabel(f"{lab}\nactivation", fontsize=9)
            else: ax.tick_params(labelleft=False)
            if r == 1: ax.set_xlabel("$x$", fontsize=9.5)
    fig.suptitle(f"Neuron-by-neuron comparison in {layer} — {label_a}  vs  {label_b}",
                 fontsize=12, color="#14285a", weight="bold", y=1.02)
    plt.tight_layout()
    plt.savefig(os.path.join(CKPT_ROOT, "figures", f"{fname}.png"), dpi=160,
                bbox_inches="tight")
    plt.show()

compare_neuron_profiles("B3_q5_linear_expval", "B6_q5_none_expval",
                        "linear entanglement", "no entanglement",
                        layer="L1", fname="N2_entanglement_5q")
```

"no dead neurons, no saturation". This reproduces that diagnostic for the QAPINN, so the two are
directly comparable in the final report.

```python
def plot_neuron_histograms(tag, fname=None):
    if tag not in ACTS:
        print(f"[skip] {tag}"); return
    acts = ACTS[tag]
    layers = list(acts.keys())
    ncol = min(3, len(layers))
    nrow = int(np.ceil(len(layers)/ncol))
    fig, axs = plt.subplots(nrow, ncol, figsize=(4.3*ncol, 2.9*nrow), squeeze=False)
    for i, L in enumerate(layers):
        ax = axs[i//ncol, i % ncol]
        A = acts[L].ravel()
        ax.hist(A, bins=45, color="#14285a", alpha=0.85, edgecolor="white", linewidth=0.4)
        ax.set_title(f"{L}   ({acts[L].shape[1]} neurons)", fontsize=10,
                     color="#14285a", weight="bold")
        ax.set_xlabel("activation value", fontsize=9.5)
        ax.set_ylabel("count", fontsize=9.5)
        ax.grid(alpha=0.2)
        dead = float(np.mean(np.abs(acts[L]).max(axis=0) < 1e-3)*100)
        sat  = float(np.mean(np.abs(A) > 0.99)*100)
        ax.text(0.03, 0.95, f"dead: {dead:.0f}%\nsaturated: {sat:.0f}%",
                transform=ax.transAxes, fontsize=8, va="top",
                bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="#cccccc", alpha=0.9))
    for j in range(len(layers), nrow*ncol):
        axs[j//ncol, j % ncol].axis("off")
    fig.suptitle(f"Neuron activation histograms — {tag}\n"
                 f"(healthy = broad spread, no spike at 0, no pile-up at $\\pm1$)",
                 fontsize=12, color="#14285a", weight="bold", y=1.02)
    plt.tight_layout()
    plt.savefig(os.path.join(CKPT_ROOT, "figures",
                             fname or f"N4_histograms_{tag}.png"), dpi=160, bbox_inches="tight")
    plt.show()

for cand in ["B3_q5_linear_expval", "B2_q4_linear_expval", "B1_q3_linear_expval"]:
    if cand in ACTS:
        plot_neuron_histograms(cand); break
```

## Figure N5 : Per-neuron frequency content across the ablation grid



```python
def spectral_centroid(profile):
    """Mean frequency weighted by power, for one neuron's activation profile."""
    p = profile - profile.mean()
    P = np.abs(np.fft.rfft(p))**2
    f = np.fft.rfftfreq(len(p), d=(2.0/len(p)))
    tot = P.sum()
    return float((f*P).sum()/tot) if tot > 0 else 0.0

def centroid_table(layer="L1"):
    rows = []
    for tag, nq, nl, ent, meas in ABLATION_GRID:
        if tag not in ACTS or layer not in ACTS[tag]:
            continue
        A = ACTS[tag][layer]
        cs = [spectral_centroid(A[:, j]) for j in range(A.shape[1])]
        rows.append((tag, nq, ent, meas, float(np.mean(cs)), float(np.std(cs)), cs))
    return rows

rows = centroid_table("L1")
if rows:
    print(f"{'run':26s} {'q':>2} {'entang.':>8} {'meas':>7} {'mean centroid':>14} {'std':>7}")
    print("-"*74)
    for tag, nq, ent, meas, m, s, _ in rows:
        print(f"{tag:26s} {nq:>2} {ent:>8} {meas:>7} {m:>14.3f} {s:>7.3f}")

    # figure: qubit sweep (linear/expval) + entanglement contrast
    base = [r for r in rows if r[2] == "linear" and r[3] == "expval"]
    none = [r for r in rows if r[2] == "none"   and r[3] == "expval"]
    fig, ax = plt.subplots(1, 2, figsize=(13, 4.3))

    a = ax[0]
    if base:
        a.errorbar([r[1] for r in base], [r[4] for r in base],
                   yerr=[r[5] for r in base], fmt="o-", lw=2.2, ms=8,
                   capsize=4, color="#14285a", label="linear entanglement")
    if none:
        a.errorbar([r[1] for r in none], [r[4] for r in none],
                   yerr=[r[5] for r in none], fmt="s--", lw=2.2, ms=8,
                   capsize=4, color="#9aa3b0", label="no entanglement")
    a.set_xlabel("qubits", fontsize=10)
    a.set_ylabel("mean spectral centroid of $L1$ neurons", fontsize=10)
    a.set_title("(a) Frequency content vs qubit count", fontsize=11,
                color="#14285a", weight="bold")
    a.grid(alpha=0.25); a.legend(fontsize=8.5)
    if base: a.set_xticks([r[1] for r in base])

    a = ax[1]
    pick = [r for r in rows if r[0] in ("B3_q5_linear_expval", "B6_q5_none_expval",
                                        "B10_q5_all_expval")]
    if pick:
        labels = {"B3_q5_linear_expval": "linear", "B6_q5_none_expval": "none",
                  "B10_q5_all_expval": "all-to-all"}
        for i, r in enumerate(pick):
            a.scatter([i]*len(r[6]), r[6], s=26, color="#6a3d9a", alpha=0.75, zorder=3)
            a.scatter([i], [r[4]], s=130, marker="_", color="#14285a", zorder=4, linewidths=2.5)
        a.set_xticks(range(len(pick)))
        a.set_xticklabels([labels.get(r[0], r[0]) for r in pick], fontsize=9.5)
        a.set_ylabel("spectral centroid (per neuron)", fontsize=10)
        a.set_xlabel("entanglement topology (5 qubits)", fontsize=10)
        a.set_title("(b) Per-neuron spread by entanglement\n(dot = neuron, bar = mean)",
                    fontsize=11, color="#14285a", weight="bold")
        a.grid(alpha=0.25, axis="y")
    plt.tight_layout()
    plt.savefig(os.path.join(CKPT_ROOT, "figures", "N5_spectral_centroid.png"),
                dpi=160, bbox_inches="tight")
    plt.show()

    np.savez(os.path.join(CKPT_ROOT, "results", "B_neuron_centroids.npz"),
             tags=np.array([r[0] for r in rows]),
             qubits=np.array([r[1] for r in rows]),
             mean_centroid=np.array([r[4] for r in rows]),
             std_centroid=np.array([r[5] for r in rows]))
    print("\n[saved] results/B_neuron_centroids.npz")
else:
    print("[skip] no activations available yet")
```
